# Layer Normalization In Transformer

```python
import torch
import torch.nn as nn
```

```python
class LayerNorm(nn.Module):
  def __init__(self, d_model, eps=1e-6 ):
    super().__init__()
    self.gamma = nn.Parameter(torch.ones(d_model))
    self.beta = nn.Parameter(torch.zeros(d_model))
    self.eps = eps

  def forward(self, x):
    mean = x.mean(dim=-1, keepdim=True)
    var = x.var(dim=-1, keepdim=True, unbiased=False)

    #Normalize the input
    norm = (x - mean) / torch.sqrt(var + self.eps)

    # Apply Scailing and Shifting
    return self.gamma * norm + self.beta
```

```python
import torch
import torch.nn as nn

d_model = 8
batch, seq_len = 2, 5

x = torch.randn(batch, seq_len, d_model)
norm = nn.LayerNorm(d_model)
out  = norm(x)

print("Input  shape:", x.shape)    # (2, 5, 8)
print("Output shape:", out.shape)  # (2, 5, 8)

# After normalization — each token vector should have mean≈0, std≈1

print("\nOne token after norm:")
print("Mean:", out[0, 0].mean().item())
print("Std: ", out[0, 0].std().item())

# Learnable parameters
print("\nGamma (initial):", norm.weight)
print("Beta  (initial):", norm.bias)
```
